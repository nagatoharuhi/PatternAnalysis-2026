# Step 3: Define Image Transforms

Resize grayscale MRI images to 128 × 128 pixels, convert them into PyTorch tensors, and normalize pixel intensities for CNN training. The same preprocessing will be applied to the GFNet model for a fair comparison.

```python
# # Set image size
# image_size = 128

# # Define image preprocessing
# image_transform = transforms.Compose([
#     transforms.Resize((image_size, image_size)),
#     transforms.ToTensor(),
#     transforms.Normalize(mean=[0.5], std=[0.5])
# ])

# print("Image transform:")
# print(image_transform)

image_size = 256

# Validation and test preprocessing
image_transform = transforms.Compose([
    transforms.Resize((image_size, image_size)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5], std=[0.5])
])

# Training-only augmentation
train_transform = transforms.Compose([
    transforms.Resize((image_size, image_size)),

    transforms.RandomAffine(
        degrees=10,
        translate=(0.05, 0.05),
        scale=(0.98, 1.02),
        interpolation=transforms.InterpolationMode.BILINEAR,
        fill=0
    ),

    transforms.ToTensor(),

    transforms.RandomErasing(
        p=0.4,
        scale=(0.01, 0.10),
        ratio=(0.5, 2.0),
        value=0
    ),

    transforms.RandomApply([
        transforms.ElasticTransform(
            alpha=10.0,
            sigma=3.0,
            interpolation=transforms.InterpolationMode.BILINEAR,
            fill=0
        )
    ], p=0.3),

    transforms.Normalize(mean=[0.5], std=[0.5])
])

print("Training transforms:")
print(train_transform)
print("\nValidation/test transforms:")
print(image_transform)
```

# Step 4: Create PyTorch Dataset

Define a custom PyTorch Dataset to load MRI images from the CSV records, apply image preprocessing, and convert AD/NC labels into numerical values for binary classification.

```python
class MRIDataset(Dataset):

    def __init__(self, dataframe, transform=None):

        self.dataframe = dataframe.reset_index(drop=True)
        self.transform = transform

        # Convert class labels to numbers
        self.label_map = {
            "NC": 0,
            "AD": 1
        }

    def __len__(self):

        return len(self.dataframe)

    def __getitem__(self, idx):

        # Get one image record
        record = self.dataframe.iloc[idx]

        # Load image as grayscale
        with Image.open(record["local_path"]) as source:
          image = source.convert("L")

        # Apply preprocessing
        if self.transform is not None:
            image = self.transform(image)

        # Convert label to integer
        label = self.label_map[record["label"]]

        return image, label
```

# Step 6: Build CNN Model

Define a CNN baseline with three convolutional blocks for extracting spatial features from grayscale MRI images. Global average pooling and a fully connected layer are used to predict the two classes: Normal Control (NC) and Alzheimer's Disease (AD).

```python
class CNNBaseline(nn.Module):

    def __init__(self):
        super().__init__()

        # Convolutional feature extraction
        self.features = nn.Sequential(

            # Block 1
            nn.Conv2d(1, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2),

            # Block 2
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2),

            # Block 3
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2)
        )

        # Global average pooling
        self.pool = nn.AdaptiveAvgPool2d((1, 1))

        # Classification layer
        self.classifier = nn.Linear(128, 2)

    def forward(self, x):

        x = self.features(x)
        x = self.pool(x)
        x = torch.flatten(x, 1)
        x = self.classifier(x)

        return x
```

# Step 7: Train CNN Model

Train the CNN baseline using cross-entropy loss and the Adam optimizer. Record training loss and accuracy for each epoch to monitor model learning. The test set is reserved for final evaluation.

```python
# # Set training parameters
# epochs = 30
# learning_rate = 0.001

# # Define loss function
# criterion = nn.CrossEntropyLoss()

# # Define optimizer
# optimizer = optim.Adam(
#     model.parameters(),
#     lr=learning_rate
# )

# # Store training history
# train_loss_history = []
# train_acc_history = []

# print("Epochs:", epochs)
# print("Learning rate:", learning_rate)
# print("Loss function:", criterion)
# print("Optimizer: Adam")

# Reset seeds and initialize a fresh CNN
seed = 42
random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(seed)

model = CNNBaseline().to(device)

epochs = 10
learning_rate = 0.001

criterion = nn.CrossEntropyLoss()

optimizer = optim.Adam(
    model.parameters(),
    lr=learning_rate
)

train_loss_history = []
train_acc_history = []
val_loss_history = []
val_acc_history = []

best_val_loss = float("inf")
best_epoch = 0

checkpoint_path = (
    "/content/drive/MyDrive/COMP3710/project/"
    "models/cnn_baseline_best.pth"
)
os.makedirs(os.path.dirname(checkpoint_path), exist_ok=True)

print("Epochs:", epochs)
print("Learning rate:", learning_rate)
print("Optimizer: Adam")
print("Checkpoint:", checkpoint_path)
```

```python
# # Train CNN model
# for epoch in range(epochs):

#     # Set model to training mode
#     model.train()

#     total_loss = 0.0
#     correct = 0
#     total = 0

#     for images, labels in train_loader:

#         # Move images and labels to GPU
#         images = images.to(device)
#         labels = labels.to(device)

#         # Clear previous gradients
#         optimizer.zero_grad()

#         # Forward pass
#         outputs = model(images)

#         # Compute loss
#         loss = criterion(outputs, labels)

#         # Backward pass
#         loss.backward()

#         # Update model parameters
#         optimizer.step()

#         # Accumulate loss weighted by batch size
#         total_loss += loss.item() * images.size(0)

#         # Get predicted classes
#         predictions = torch.argmax(outputs, dim=1)

#         # Count correct predictions
#         correct += (predictions == labels).sum().item()
#         total += labels.size(0)

#     # Calculate average loss and accuracy
#     epoch_loss = total_loss / total
#     epoch_acc = correct / total

#     # Save training history
#     train_loss_history.append(epoch_loss)
#     train_acc_history.append(epoch_acc)

#     print(
#         f"Epoch [{epoch + 1}/{epochs}] "
#         f"Loss: {epoch_loss:.4f} "
#         f"Accuracy: {epoch_acc * 100:.2f}%"
#     )

for epoch in range(epochs):

    # 1. Training
    model.train()
    train_total_loss = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in train_loader:
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        train_total_loss += loss.item() * labels.size(0)
        train_correct += (
            outputs.argmax(dim=1) == labels
        ).sum().item()
        train_total += labels.size(0)

    train_loss = train_total_loss / train_total
    train_acc = train_correct / train_total

    train_loss_history.append(train_loss)
    train_acc_history.append(train_acc)

    # 2. Validation
    model.eval()
    val_total_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for images, labels in val_loader:
            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            val_total_loss += loss.item() * labels.size(0)
            val_correct += (
                outputs.argmax(dim=1) == labels
            ).sum().item()
            val_total += labels.size(0)

    val_loss = val_total_loss / val_total
    val_acc = val_correct / val_total

    val_loss_history.append(val_loss)
    val_acc_history.append(val_acc)

    # 3. Save the best validation model
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_epoch = epoch + 1

        torch.save(model.state_dict(), checkpoint_path)
        print(f"Best model saved at Epoch {best_epoch}")

    print(
        f"Epoch [{epoch + 1}/{epochs}] "
        f"Train Loss: {train_loss:.4f} "
        f"Train Acc: {train_acc:.2%} | "
        f"Val Loss: {val_loss:.4f} "
        f"Val Acc: {val_acc:.2%}",
        flush=True
    )

print("\nTraining completed.")
print("Best epoch:", best_epoch)
print("Best validation loss:", round(best_val_loss, 4))
print("Checkpoint:", checkpoint_path)
```

```python
# Analyse prediction confidence on the test set
import numpy as np
import pandas as pd

# labels = np.asarray(roc_labels, dtype=int)
# ad_probs = np.asarray(roc_probs, dtype=float)

labels = np.asarray(all_labels, dtype=int)
ad_probs = np.asarray(all_probabilities, dtype=float)

assert len(labels) == len(ad_probs) == len(test_dataset)
assert np.isfinite(ad_probs).all()
assert ((ad_probs >= 0) & (ad_probs <= 1)).all()

# Class mapping: NC = 0, AD = 1
# At an exact tie, choose NC to match argmax over [P(NC), P(AD)]
predictions = (ad_probs > 0.5).astype(int)

# Confidence in the predicted class
confidence = np.maximum(ad_probs, 1 - ad_probs)
correct = predictions == labels

confidence_results = pd.DataFrame({
    "true_label": labels,
    "predicted_label": predictions,
    "ad_probability": ad_probs,
    "confidence": confidence,
    "correct": correct
})

summary_rows = []

for group_name, mask in [
    ("Correct predictions", correct),
    ("Incorrect predictions", ~correct)
]:
    values = confidence[mask]

    summary_rows.append({
        "Group": group_name,
        "Count": len(values),
        "Mean confidence": values.mean() if len(values) else np.nan,
        "Median confidence": np.median(values) if len(values) else np.nan
    })

# print("GFNet Test Confidence Summary")
print("CNN Test Confidence Summary")
print(pd.DataFrame(summary_rows).round(4).to_string(index=False))

# Descriptive diagnostic only; this is not a selected referral threshold
high_confidence = confidence >= 0.90
high_confidence_errors = high_confidence & ~correct

print("\nPredictions with confidence >= 0.90:",
      int(high_confidence.sum()))
print("Incorrect predictions with confidence >= 0.90:",
      int(high_confidence_errors.sum()))

if high_confidence.any():
    print(
        "Accuracy among predictions with confidence >= 0.90:",
        f"{correct[high_confidence].mean():.2%}"
    )
```

```python
import numpy as np
import pandas as pd
import torch

# Reload the checkpoint selected by validation loss.
model.load_state_dict(
    torch.load(
        checkpoint_path,
        map_location=device,
        weights_only=True
    )
)
model.eval()

def collect_predictions(loader):
    labels_list = []
    predictions_list = []
    confidence_list = []

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(device)

            probabilities = torch.softmax(model(images), dim=1)
            confidence, predictions = probabilities.max(dim=1)

            labels_list.extend(labels.cpu().tolist())
            predictions_list.extend(predictions.cpu().tolist())
            confidence_list.extend(confidence.cpu().tolist())

    return pd.DataFrame({
        "true_label": labels_list,
        "prediction": predictions_list,
        "confidence": confidence_list
    })

# Select the threshold using validation data only.
validation_predictions = collect_predictions(val_loader)

target_accuracy = 0.90
minimum_coverage = 0.10

# Fixed candidate grid: 0.50, 0.51, ..., 0.99
thresholds = np.arange(50, 100) / 100

rows = []

for threshold in thresholds:
    retained = (
        validation_predictions["confidence"] >= threshold
    )
    retained_count = int(retained.sum())

    if retained_count:
        retained_accuracy = (
            validation_predictions.loc[retained, "prediction"]
            == validation_predictions.loc[retained, "true_label"]
        ).mean()
    else:
        retained_accuracy = np.nan

    rows.append({
        "threshold": threshold,
        "retained_count": retained_count,
        "coverage": retained.mean(),
        "retained_accuracy": retained_accuracy
    })

threshold_results = pd.DataFrame(rows)

eligible = threshold_results[
    (threshold_results["coverage"] >= minimum_coverage)
    & (threshold_results["retained_accuracy"] >= target_accuracy)
]

print("CNN validation threshold analysis")
print(threshold_results.round(4).to_string(index=False))

selected_threshold = None

if eligible.empty:
    print(
        "\nNo candidate threshold met both validation criteria: "
        "accuracy >= 90% and coverage >= 10%."
    )
else:
    # Maximise coverage; break ties with the lower threshold.
    selected = eligible.sort_values(
        ["coverage", "threshold"],
        ascending=[False, True]
    ).iloc[0]

    selected_threshold = float(selected["threshold"])

    print(f"\nSelected threshold: {selected_threshold:.2f}")
    print(f"Validation coverage: {selected['coverage']:.2%}")
    print(
        "Validation retained accuracy:",
        f"{selected['retained_accuracy']:.2%}"
    )

    # Freeze the threshold before evaluating test performance.
    test_predictions = collect_predictions(test_loader)

    retained = (
        test_predictions["confidence"] >= selected_threshold
    )
    correct = (
        test_predictions["prediction"]
        == test_predictions["true_label"]
    )
    is_ad = test_predictions["true_label"] == 1
    is_nc = test_predictions["true_label"] == 0

    retained_count = int(retained.sum())
    retained_errors = int((retained & ~correct).sum())

    # Errors remaining among predictions not referred.
    retained_fn = int((
        retained & is_ad & (test_predictions["prediction"] == 0)
    ).sum())
    retained_fp = int((
        retained & is_nc & (test_predictions["prediction"] == 1)
    ).sum())

    print("\nCNN test referral results")
    print("Test slices:", len(test_predictions))
    print("Retained slices:", retained_count)
    print("Referred slices:", int((~retained).sum()))
    print(f"Coverage: {retained.mean():.2%}")
    print(f"Referral rate: {(~retained).mean():.2%}")

    if retained_count:
        print(
            "Retained accuracy:",
            f"{correct[retained].mean():.2%}"
        )
    else:
        print("Retained accuracy: undefined (no retained slices)")

    print("Retained errors:", retained_errors)
    print("Retained false negatives:", retained_fn)
    print("Retained false positives:", retained_fp)

    print(f"AD coverage: {retained[is_ad].mean():.2%}")
    print(f"NC coverage: {retained[is_nc].mean():.2%}")
```

```python
import gc
import numpy as np
import pandas as pd
import torch

# Change to "GFNet" in the GFNet notebook.
model_name = "CNN"

assert torch.cuda.is_available(), "This benchmark requires a CUDA GPU."

benchmark_device = next(model.parameters()).device
assert benchmark_device.type == "cuda"
assert next(model.parameters()).dtype == torch.float32

model.eval()

# Remove stored parameter gradients without changing model weights.
model.zero_grad(set_to_none=True)

gc.collect()
torch.cuda.empty_cache()

parameter_count = sum(p.numel() for p in model.parameters())

# Fixed input shape, already on GPU.
benchmark_input = torch.randn(
    1, 1, 256, 256,
    device=benchmark_device,
    dtype=torch.float32
)

warmup_runs = 30
timed_runs = 200

with torch.inference_mode():
    # Warm up kernels and FFT plans.
    for _ in range(warmup_runs):
        output = model(benchmark_input)

    del output
    torch.cuda.synchronize(benchmark_device)

    # Measure GPU forward-pass latency using CUDA events.
    times_ms = []

    for _ in range(timed_runs):
        start = torch.cuda.Event(enable_timing=True)
        end = torch.cuda.Event(enable_timing=True)

        start.record()
        output = model(benchmark_input)
        end.record()

        end.synchronize()
        times_ms.append(start.elapsed_time(end))
        del output

    # Measure memory separately from timing.
    torch.cuda.synchronize(benchmark_device)
    torch.cuda.reset_peak_memory_stats(benchmark_device)

    baseline_bytes = torch.cuda.memory_allocated(benchmark_device)

    output = model(benchmark_input)
    torch.cuda.synchronize(benchmark_device)

    peak_bytes = torch.cuda.max_memory_allocated(benchmark_device)
    incremental_bytes = peak_bytes - baseline_bytes

    del output

mib = 1024 ** 2

benchmark_results = pd.DataFrame([{
    "Model": model_name,
    "Parameters": parameter_count,
    "Batch size": 1,
    "Mean latency (ms)": np.mean(times_ms),
    "Median latency (ms)": np.median(times_ms),
    "P95 latency (ms)": np.percentile(times_ms, 95),
    "Baseline allocated (MiB)": baseline_bytes / mib,
    "Peak allocated (MiB)": peak_bytes / mib,
    "Incremental peak (MiB)": incremental_bytes / mib
}])

print("GPU:", torch.cuda.get_device_name(benchmark_device))
print("PyTorch:", torch.__version__)
print("CUDA:", torch.version.cuda)
print("cuDNN benchmark:", torch.backends.cudnn.benchmark)
print("cuDNN TF32:", torch.backends.cudnn.allow_tf32)
print("Matmul TF32:", torch.backends.cuda.matmul.allow_tf32)
print("\nForward-only inference benchmark")
print(benchmark_results.round(3).to_string(index=False))

del benchmark_input
```
